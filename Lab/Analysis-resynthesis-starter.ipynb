### Helpers

Two small functions to save some typing.
`show_spec` displays a spectrogram in dB with the right time and frequency axes.
`listen` plays several signals, each under its own label.

```python
def show_spec(S_db, title, y_axis="hz"):
    """Display a dB spectrogram (bins x frames). Use y_axis="mel" for mel bands."""
    fig, ax = plt.subplots(figsize=(9, 3.5))
    img = librosa.display.specshow(
        S_db, sr=FS, hop_length=hop_len, x_axis="time", y_axis=y_axis, ax=ax
    )
    fig.colorbar(img, ax=ax, format="%+2.0f dB")
    ax.set_title(title)
    fig.tight_layout()
    return fig


def listen(**signals):
    """Play each signal under its label, e.g., listen(original=xs, rebuilt=ys)."""
    for label, ys in signals.items():
        print(label)
        display(Audio(ys, rate=FS))
```
